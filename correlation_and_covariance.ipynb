Q1. Define Covariance and explain how it differs from Correlation in terms of scale and interpretation

Ans. Covariance indicates the direction of the relationship between two variables, while correlation standardizes covariance to show both direction and strength on a fixed scale from −1 to +1.

Q2.  What does a positive, negative, and zero covariance indicate about the relationship between two variables?

Ans. Positive covariance means variables move together, negative covariance means they move in opposite directions, and zero covariance indicates no linear relationship between them.

Q3. . Discuss the limitations of covariance as a measure of relationship between two variables. Why is correlation
preferred in many cases?

Ans. Covariance has limitations because it depends on measurement units, has no fixed scale, and does not indicate the strength of a relationship. Correlation is preferred as it standardizes covariance, providing a unit-free measure with a fixed range from −1 to +1, making interpretation and comparison easier.

Q4. Explain the difference between Pearson’s correlation coefficient and Spearman’s rank correlation
coefficient. When would you prefer to use Spearman’s correlation?




Ans. Pearson’s correlation measures linear relationships using actual values, while Spearman’s correlation measures monotonic relationships using ranks. Spearman’s correlation is preferred when data is ordinal, non-normal, contains outliers, or the relationship is not linear.

Q5. If the correlation coefficient between two variables X and Y is 0.85, interpret this value in context. Can you
infer causation from this value? Why or why not?

Ans. A correlation coefficient of 0.85 indicates a strong positive linear relationship between X and Y. However, causation cannot be inferred because correlation does not establish cause-and-effect and may be influenced by reverse causality or confounding variables.
